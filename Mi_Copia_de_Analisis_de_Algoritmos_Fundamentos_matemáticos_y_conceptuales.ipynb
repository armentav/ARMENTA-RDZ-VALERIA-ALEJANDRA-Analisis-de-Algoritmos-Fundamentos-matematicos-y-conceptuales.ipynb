- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  
El Principio de Invarianza establece que el tiempo de ejecución de dos implementaciones distintas de un mismo algoritmo no difiere más que en una constante multiplicativa.  
Formalmente, si se tiene un algoritmo y dos de sus implementaciones, $I_1$ e $I_2$ (que pueden ser códigos distintos o ejecutarse en computadoras distintas), cuyos tiempos de ejecución respectivos son $T_1(n)$ y $T_2(n)$, el principio afirma que existe una constante real $c > 0$ y un número natural $n_0$ tales que para todo tamaño de entrada $n \geq n_0$ se verifica la siguiente relación:  
$$T_1(n) \le c \cdot T_2(n)$$De este principio se derivan las siguientes conclusiones fundamentales para el análisis y diseño de algoritmos:  


* Independencia de la implementación: El tiempo necesario para resolver un problema mediante un algoritmo depende de la naturaleza intrínseca del propio algoritmo y no de factores externos o de la implementación concreta (como el lenguaje de programación, el compilador, la habilidad del programador o la máquina física en la que se ejecuta).
* Mismo orden de complejidad: Debido a que las diferencias se limitan a un factor constante, dos implementaciones "razonables" de un mismo algoritmo compartirán siempre el mismo orden de complejidad asintótica. Por ejemplo, si una implementación del método de la burbuja es de orden cuadrático ($O(n^2)$), cualquier otra implementación del mismo algoritmo también será de orden cuadrático ($O(n^2)$).  
* La constante oculta: Aunque el orden de complejidad es idéntico, la constante multiplicativa que relaciona ambas implementaciones (a menudo denominada constante oculta) puede ser decisiva en la práctica de la ingeniería de software para determinar cuál de ellas se ejecuta más rápido ante conjuntos de datos reales de tamaño moderado.

**RESPUESTA:**
1. ¿Qué es un algoritmo y cuáles son sus características clave? Un **algoritmo** es una secuencia ordenada y finita de instrucciones precisas diseñadas para resolver un problema o transformar datos de entrada en una salida deseada. **Características clave:** * **Entrada:** Recibe datos iniciales para operar. * **Salida:** Produce un resultado tras procesar la entrada. * **Precisión:** Cada paso es claro y sin ambigüedades. * **Finitud:** Debe terminar tras un número limitado de pasos y en tiempo finito. * **Determinismo:** Con las mismas entradas, siempre produce la misma salida. * **Efectividad:** Cada instrucción es básica y realizable en tiempo finito. * **Corrección y generalidad:** Resuelve el problema correctamente para cualquier caso válido.
2. El Problema de Paro (*Halting Problem*) Formulado por **Alan Turing (1936)**, plantea si es posible crear un algoritmo capaz de analizar cualquier otro programa y sus datos de entrada para predecir con certeza si terminará de ejecutarse o si caerá en un ciclo infinito. Turing demostró matemáticamente que **no existe tal algoritmo**, estableciendo que es un problema **indecidible** (no computable).
3. Análisis a Priori vs. Prueba a Posteriori * **Análisis a Priori (Teórico):** Se realiza matemáticamente *antes* de programar. Da como resultado una función que estima la eficiencia de forma **independiente** de la computadora, el lenguaje o el compilador. * **Prueba a Posteriori (Experimental):** Consiste en programar y ejecutar el algoritmo en una computadora real para medir el tiempo y la memoria consumidos. Sus resultados **dependen** directamente del hardware, el lenguaje, el compilador y los datos de prueba.


- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**
Cuando el tiempo de ejecución de un algoritmo depende no solo del tamaño de los datos ($n$), sino también de cómo vienen organizados, su eficiencia se evalúa en tres escenarios:
1. Peor caso (Worst-case): Es la entrada de tamaño $n$ con la que el algoritmo tarda más tiempo o ejecuta la mayor cantidad de operaciones básicas. Su análisis es el más utilizado porque establece una cota superior que garantiza que el algoritmo nunca tardará más que ese límite.

Ejemplo: En una búsqueda secuencial, ocurre cuando el elemento buscado está al final de la lista o no existe en ella.

2. Mejor caso (Best-case): Es la entrada de tamaño $n$ con la que el algoritmo tarda menos tiempo o ejecuta el menor número de operaciones básicas. No significa una entrada pequeña, sino la entrada más favorable para ese tamaño $n$. Aunque no garantiza el rendimiento general, sirve para descartar algoritmos lentos incluso en condiciones ideales o para aprovechar datos casi ordenados.

Ejemplo: En una búsqueda secuencial, ocurre cuando el elemento buscado se encuentra en la primera posición.

3. Caso promedio (Average-case): Representa el rendimiento esperado ante una entrada "típica" o aleatoria de tamaño $n$, ponderando el costo de cada posible entrada según su probabilidad de ocurrir. Es el más difícil de calcular porque requiere conocer o asumir la distribución de probabilidad de los datos, y no equivale simplemente al promedio entre el mejor y el peor caso.

Ejemplo: En un algoritmo de ordenamiento, corresponde a una lista cuyos elementos están desordenados de forma aleatoria.

## 4. Sección Práctica 2: Iteración vs. Recursión
Un algoritmo recursivo debe llamarse a sí mismo resolviendo un problema menor e incluir siempre un "caso base" para evitar ciclos infinitos.

### [CÓDIGO 2.1] Sucesión de Fibonacci Recursiva
```python
# Fi = Fi-1 + Fi-2 para i >= 2; F0 = 0, F1 = 1
def fibonacci_recursivo(n):
    if n == 0: return 0
    elif n == 1: return 1
    else: return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)

````python
# Fi = Fi-1 + Fi-2 para i >= 2; F0 = 0, F1 = 1
def fibonacci_recursivo(n):
    if n == 0: return 0
    elif n == 1: return 1
    else: return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)
````
